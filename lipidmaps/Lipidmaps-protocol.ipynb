converting a json file into a csv file

```python
from rdkit import Chem
from rdkit.Chem import PandasTools
import pandas as pd

# Load SDF file
sdf_file = 'SECRETed_space_Feb23.sdf'  # replace with the path to your SDF file
suppl = Chem.SDMolSupplier(sdf_file)

# Convert SDF to Pandas DataFrame
data = []
for mol in suppl:
    if mol is not None:
        data.append(mol)
df = PandasTools.LoadSDF(sdf_file)
df = pd.DataFrame(df)

# Save DataFrame to CSV
csv_file = 'SECRETed_space_Feb23_description.csv'  # replace with the desired name for your CSV file
df.to_csv(csv_file, index=False)
print(f"CSV file '{csv_file}' saved successfully!")
```

Taxonomydownloader

```python
import requests
import csv

# Load CSV file with download links and PubChem CIDs
with open('LOTUSSimilar-links.csv', newline='') as file:
    reader = csv.reader(file)
    next(reader)  # Skip header row
    for row in reader:
        download_link = row[0]
        cid = row[1]
        # Send GET request to download link
        response = requests.get(download_link)
        
        # Check if response is successful
        if response.status_code == 200:
            # Save downloaded content as CSV file
            with open(f'CID_{cid}_LOTUSSimilar-taxonomy.csv', 'wb') as f:
                f.write(response.content)
            print(f"CSV file for CID {cid} downloaded successfully!")
        else:
            print(f"Failed to download CSV file for CID {cid} with status code {response.status_code}")
```
