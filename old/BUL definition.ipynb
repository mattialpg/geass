```python
import pickle
import pandas as pd
from rdkit.Chem.PandasTools import LoadSDF
```

```python
df_lipids = LoadSDF('data/lipid_structures.sdf', smilesName='SMILES', molColName='Molecule', strictParsing=True)
```

```python
df_lipids.count()
```

```python
# Get the rows that DON'T have PUBCHEM_CID
df_lipids_NO_pubchem = df_lipids[df_lipids['PUBCHEM_CID'].isna()]
```

```python
# Get the rows that have PUBCHEM_CID
df_lipids_pubchem = df_lipids[df_lipids['PUBCHEM_CID'].notna()]
```

```python
lipid_mols = {cid: molecule for cid, molecule in zip(df_lipids_pubchem['PUBCHEM_CID'], df_lipids_pubchem['Molecule'])}
lipid_mols.update({id: molecule for id, molecule in zip(df_lipids_NO_pubchem['LM_ID'], df_lipids_NO_pubchem['Molecule'])})
```

```python
lipid_mols
```

```python
with open('backups/lipid_mols.pkl', 'wb') as output_file:
    pickle.dump(lipid_mols, output_file)
```

```python
with open('../recap-delivery/data/BUL_names.pkl', 'rb') as f:
    BULnames = pickle.load(f)
```

```python
BULnamesv2 = {}
for names, mol in BULnames.items():
    names_no_mod = tuple(n for n in names if 'mod' not in n)
    if names_no_mod:
        BULnamesv2[names_no_mod] = mol
```

```python
lipid_mols_with_tuples = {(code,): mol for code, mol in lipid_mols.items()}
```

```python
BUL_with_lipids = BULnamesv2.copy()
for names, mol in lipid_mols_with_tuples.items():
    if not any(names[0] in k for k in BUL_with_lipids.keys()):
        BUL_with_lipids[names] = mol
```

```python
with open('data/BUL_with_lipids.pkl', 'wb') as output_file:
    pickle.dump(BUL_with_lipids, output_file)
```
